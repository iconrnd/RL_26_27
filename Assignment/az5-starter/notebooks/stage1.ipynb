## (e) Monte Carlo evaluation

$v_{\mathrm{rand}}(s)$ for the positions below, from $n$ = 100, 1 000, and 10 000 games, with 95 % intervals, and the dependence of the interval width on $n$.

```python
# Positions supplied for the Monte Carlo evaluation (moves from the empty board).
POSITIONS = {
    "empty board": [],
    "P1": ["C3"],
    "P2": ["C3", "A1"],
    "P3": ["C3", "C4", "D4", "B4", "B3", "C5"],
}
states = {}
for name, moves in POSITIONS.items():
    s = GoState()
    for m in moves:
        s = s.play(parse_move(m))
    states[name] = s
    print(name, "- to move:", "Black" if s.to_play == 1 else "White")
    print(s)
```
